#5.1 합성곱 신경망

신경망은 많은 연산량을 요구함. 이런 문제를 해결하기 위해 합성곱 신경망을 사용함. 합성곱 신경망은 이미지 전체를 한 번에 계산하는 것이 아니라 이미지의 국소적인 부분을 계산함으로써 시간과 자원을 절약함

##5.1.1 합성곱층의 필요성

이미지, 영상을 처리하는데 유용함



일반적인 이미지 분석은 이미지 데이터의 공간적인 구조를 무시하고 그냥 벡터로 처리함.

##5.1.2 합성곱 신경망 구조

1. 입력층
2. 합성곱층
3. 풀링층
4. 완전연결층
5. 출력층

-합성곱 신경망은 합성곱층과 풀링층을 거치면서 입력 이미지의 주요 특성 벡터를 추출함. 그 후 추출된 주요 특성 벡터들은 완전연결층을 거치면서 1차원 벡터로 변환되며 마지막으로 출력층에서 소프트맥스 함수를 활성화 함수로 사용하여 최종 결과 출력

**입력층**



이미지 데이터가 최초로 거치게 되는 계층


이미지는 단순 1차원 데이터가 아니라 높이, 너비, 채널*의 값을 갖는 3차원 데이터임.



*채널: 이미지가 그레이스케일이면 1의 값, 컬러면 3의 값을 가짐.


**합성곱층**


입력 데이터에서 특성을 추출하는 역할


이미지에 대한 특성을 감지하기 위해 커널*이나 필터 사용 -> 추출된 결과물: 맵


*3x3, 5x5 크기로 적용되는 것이 일반적임

<br>
<스트라이드가 1일 때 이동하는 과정>

1. 입력 이미지에 3*3 필터 적용
2. 필터가 1만큼 이동 (스트라이드 만큼)
3. 1-2 반복

<br>
<합성곱층이 하이퍼 파라미터>

- 필터 개수(K)
- 필터 크기(F)
- 스트라이드(S)
- 패딩(P)

**풀링층**

특성 맵의 차원을 다운 샘플링*하여 연산량을 감소시키고 주요한 특성 벡터를 추출하여 학습을 효과적으로 할 수 있게 함


*다운 샘플링: 이미지 크기를 축소


<br>
<풀링 연산>

- 최대 풀링: 대상 영역에서 최댓값을 추출
- 평균 풀링: 대상 영역에서 평균을 반환

<br>
<풀링층의 하이퍼파라미터>

- 필터 크기(F)
- 스트라이드(S)

**완전연결층**

3차원 벡터에서 1차원 벡터로 펼쳐짐

**출력층**

소프트맥스 활성화 함수를 사용하여 각 레이블에 속할 확률값을 출력
