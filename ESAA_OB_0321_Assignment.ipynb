```python
from sklearn.datasets import make_moons

X, y = make_moons(n_samples=1000, noise=0.05)
```

```python
from sklearn.mixture import GaussianMixture

gm = GaussianMixture(n_components=3, n_init=10)
gm.fit(X)
```

주어진 위치에서 모델의 밀도 추정 가능. score_samples()메서드는 그 위치의 확률 밀도 함수의 로그를 예측

```python
gm.score_samples(X)
```

##9.2.1 가우시안 혼합을 사용한 이상치 탐지


- 이상치 감지
  - 밀도가 낮은 지역에 있는 샘플을 이상치로 간주
: 밀도 임곗값을 지정해야 함
- 특이치 감지
  - 이상치로 오염되지 않은 데이터셋에서 훈련

```python
import numpy as np

densities = gm.score_samples(X)
density_threshold = np.percentile(densities, 4)
anomalies = X[densities < density_threshold]
```

##9.2.4 이상치 탐지와 특이치 탐지를 위한 다른 알고리즘

- PCA
- Fast_MCD
  - 혼합된 것이 아니라 하나의 가우시안 분포에서 생성되었다고 가정
- 아이솔레이션 포레스트
  - 고차원의 데이터셋에서 효율적인 알고리즘
  - 무작위로 생성한 결정 트리로 구성된 랜덤 포레스트를 만든 후 각 노드에서 특성을 랜덤하게 선택한 다음 랜덤한 임곗값을 골라 데이터셋을 둘로 나눔.
  - 모든 데이터셋이 다른 샘플과 나눠질 때까지한 후에 이상치를 탐지
  - 이상치는 일반적으로 적은 단계에서 걸러지는 경향이 있음. 이를 이용해 이상치를 탐지함
- LOF
  - 샘플 주위의 밀도와 이웃 주위의 밀도를 비교
- one-class SVM
  - 특이치 탐지에 더 적절하지만 대규모 데이터셋에서는 사용이 어려움
